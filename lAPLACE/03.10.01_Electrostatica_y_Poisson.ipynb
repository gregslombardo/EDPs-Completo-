# 3.10 · Electrostática y el problema de Poisson
**Complemento al temario oficial.** Una bola de radio $R$ contiene una densidad de carga constante $\rho$. Su frontera conductora se mantiene a potencial cero y la permitividad $\varepsilon$ es constante. La ley de Gauss $\nabla\cdot E=\rho/\varepsilon$ y $E=-\nabla\phi$ dan $-\Delta\phi=\rho/\varepsilon$, $\phi|_{\partial B_R}=0$.

### Proposición 3.10.1 · Potencial de la bola

**Complemento · Una superficie de medida dentro de la carga.** La bola contiene una distribución de carga fijada; la condición conductora se impone en su frontera. Dibuja una esfera imaginaria de radio $s<R$ para medir el flujo. Al duplicar $s$, la carga encerrada se multiplica por ocho y el área por cuatro. Por simetría, el campo normal se duplica. Así se predice $E_s\propto s$ incluso antes de calcular el potencial.

La unicidad de Dirichlet y la invariancia por rotaciones imponen simetría radial. Escribe $s=|x|$. En tres dimensiones $(s^2\phi')'=-(\rho/\varepsilon)s^2$. La regularidad en el centro excluye un término $C/s$. Imponer el dato de frontera da
$$\phi(s)=\frac{\rho}{6\varepsilon}(R^2-s^2),\qquad E(s)=\frac{\rho}{3\varepsilon}s\,e_s.$$
Así, $4\pi s^2 E_s=(4\pi/3)s^3\rho/\varepsilon$. Para $\rho>0$ el potencial es superarmónico. Su mínimo ocurre en la frontera y su máximo puede ser interior.

### Interpretación variacional
El funcional $\int_{B_R}(\varepsilon|\nabla v|^2/2-\rho v)\,dx$ sobre $H^1_0(B_R)$ tiene como ecuación de Euler–Lagrange la identidad débil de Poisson. Para la solución anterior, la diferencia de energía al sumar $w\in H^1_0$ es $\varepsilon\int|\nabla w|^2/2$, que prueba minimalidad y unicidad.

**Lectura de las dos gráficas.** En el centro la pendiente del potencial es cero por simetría. Hacia la pared el potencial baja cada vez más deprisa si $\rho>0$. La segunda gráfica da precisamente el negativo de esa pendiente. Para $R=\varepsilon=\rho=1$, a medio radio deben leerse $\phi(1/2)=1/8$ y $E_s(1/2)=1/6$.


```python
import numpy as np
import matplotlib.pyplot as plt

s = np.linspace(0,1,301)
fig, axes = plt.subplots(1,2,figsize=(8,3.4))
axes[0].plot(s,(1-s**2)/6)
axes[0].set(xlabel="Radio", ylabel="Potencial", title="Borde a tierra")
axes[1].plot(s,s/3)
axes[1].set(xlabel="Radio", ylabel="Campo radial", title="Ley de Gauss")
fig.suptitle("Bola unitaria con densidad y permitividad unitarias")
fig.tight_layout()
plt.show()
```
