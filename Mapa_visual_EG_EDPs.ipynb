```python
import matplotlib.pyplot as plt
import numpy as np
fig, axes = plt.subplots(1,3,figsize=(10.5,3.2),constrained_layout=True)
# Cauchy en la recta, c=a=1 y D=0.16, sin fuentes.
x = np.linspace(-3.2,3.2,401)
def inicial(x):
    return np.maximum(1-(x/0.72)**2,0)**3
t = 1.
onda = (inicial(x-t)+inicial(x+t))/2
y = np.linspace(-0.72,0.72,401)
gauss = np.exp(-(x[:,None]-y[None,:])**2/(4*0.16*t))/np.sqrt(4*np.pi*0.16*t)
calor = np.trapezoid(gauss*inicial(y),y,axis=1)
for ax, nombre, perfil in zip(axes,['Transporte','Onda, velocidad inicial cero','Calor'],[inicial(x-t),onda,calor]):
    ax.plot(x,inicial(x),'--',color='0.48',label='Dato inicial')
    ax.plot(x,perfil,color='#00689d',label='A tiempo uno')
    ax.set(xlabel='Posición',ylim=(-0.05,1.08),title=nombre)
axes[0].set_ylabel('Altura')
axes[-1].legend()
plt.show()
```

**Un dato inicial, tres evoluciones.** Marca el máximo del perfil inicial. En transporte se desplaza entero con $a=1$. En la onda con $c=1$ y cuerda inicialmente en reposo aparecen dos medias copias. En calor con difusividad $0.16$ se ensancha y baja. Para Laplace-Dirichlet se plantea otra pregunta: una vez fijado el borde y alcanzado el equilibrio, ¿qué valor queda en el interior?
