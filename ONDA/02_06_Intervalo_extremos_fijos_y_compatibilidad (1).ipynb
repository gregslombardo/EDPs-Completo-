# 2.6 · La cuerda con extremos fijos
**Complemento al temario del general.** Sean $c,L>0$. En $0<x<L$ buscamos $u_{tt}=c^2u_{xx}$ con $u(0,t)=u(L,t)=0$, $u(x,0)=f(x)$ y $u_t(x,0)=g(x)$.

### Proposición 2.6.1 · Modos normales
Separar $u=X(x)T(t)$ conduce a $-X''=\lambda X$ con $X(0)=X(L)=0$. Integrar contra $X$ da $\lambda\int X^2=\int (X')^2$, así que para un modo no trivial $\lambda>0$. Al resolver la EDO quedan $X_n(x)=\sin(n\pi x/L)$ y $\omega_n=cn\pi/L$.

Para $h=f,g$, escribe $h_n=(2/L)\int_0^L h(x)X_n(x)\,dx$. La proyección de la ecuación da $q_n''+\omega_n^2q_n=0$, $q_n(0)=f_n$, $q_n'(0)=g_n$. Por tanto
$$u(x,t)=\sum_{n\ge1}\left(f_n\cos(\omega_nt)+\frac{g_n}{\omega_n}\sin(\omega_nt)\right)X_n(x).$$
Para sumas finitas se verifican la EDP y los datos término a término. Una hipótesis suficiente para justificar dos derivaciones uniformes es $\sum (n^2|f_n|+n|g_n|)<\infty$, con datos continuos que coinciden con sus series seno.

### Observación 2.6.2 · Qué pasa en las esquinas
Si la solución es $C^2$ hasta las esquinas, el borde impone $f=g=0$ en ambos extremos. Derivar el borde dos veces y usar la EDP impone además $f''=0$ allí. Por ejemplo $f=x(L-x)$ satisface el primer requisito, pero $f''=-2$ contradice el último. La serie aún puede definir una solución de energía.
Para datos $C^2,C^1$ con estas compatibilidades, sus extensiones impares y $2L$-periódicas son $C^2,C^1$. Aplicar d’Alembert a las extensiones da el mismo problema y explica el cambio de signo en cada reflexión.

### Ejemplo 2.6.3 · Desplazamiento y velocidad en modos distintos
Toma $L=c=1$, $f=\sin(\pi x)$ y $g=2\pi\sin(2\pi x)$. El segundo modo arranca con desplazamiento cero y adquiere amplitud $\sin(2\pi t)$.

[Notas originales de la sección 2.6 y escena de reflexión en la cuerda](02_01_Cauchy_global_y_dAlembert%20%281%29.ipynb).

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 401)
fig, ax = plt.subplots(figsize=(8, 3.8))
for t in [0, 0.125, 0.25, 0.5]:
    u = np.sin(np.pi*x)*np.cos(np.pi*t) + np.sin(2*np.pi*x)*np.sin(2*np.pi*t)
    ax.plot(x, u, label=f"t = {t:g}")
ax.set(xlabel="Posición x", ylabel="Desplazamiento", title="Dos modos de una cuerda con extremos fijos")
ax.axhline(0, color="0.6", linewidth=0.7)
ax.legend(ncol=2)
fig.tight_layout()
plt.show()
```
