# 2.6.1. Problema mixto en un intervalo

## **Definición 2.6.1 (Problema de Dirichlet homogéneo en un intervalo).**

Sean $L>0$ y

$$
f\in C^2([0,L]),
\qquad
g\in C^1([0,L]).
$$

Se considera

$$
\begin{cases}
u_{tt}-c^2u_{xx}=0,
&x\in(0,L),\ t>0,\\
u(x,0)=f(x),
&x\in[0,L],\\
u_t(x,0)=g(x),
&x\in[0,L],\\
u(0,t)=0,
&t>0,\\
u(L,t)=0,
&t>0.
\end{cases}
$$

Los dos extremos permanecen fijos.

## **Definición 2.6.2 (Condiciones de compatibilidad).**

En las notas se escriben

$$
f(0)=f(L)=0,
$$

y

$$
g(0)=g(L)=0.
$$

Estas condiciones garantizan que los datos iniciales concuerdan con la condición de frontera en las esquinas.

### Complemento para regularidad clásica

Para que la extensión impar periódica de $f$ sea de clase $C^2$, se requiere además

$$
f''(0)=f''(L)=0.
$$

# 2.6.2. Extensión impar periódica

**Complemento · Desplegar las reflexiones.** Al duplicar un intervalo de longitud $L$ por reflexión en $L$, el punto $L+h$ representa la altura opuesta a la de $L-h$. Repetir la operación en el extremo siguiente recupera el signo tras recorrer $2L$. Así una recta $x-ct=\xi$ en el problema extendido codifica una trayectoria en zigzag con inversión en cada extremo fijo. Para $L=c=1$, una señal que sale de $x=1/4$ hacia la derecha llega a la pared en $t=3/4$ y vuelve a su punto de partida en $t=3/2$, con signo opuesto.


## **Definición 2.6.3 (Extensión impar $2L$-periódica).**

Primero definimos en $[-L,L]$

$$
\widetilde f(x)
=
\begin{cases}
f(x),&0\leq x\leq L,\\
-f(-x),&-L\leq x<0,
\end{cases}
$$

y después imponemos

$$
\widetilde f(x+2L)=\widetilde f(x).
$$

Se define $\widetilde g$ de la misma manera.

La extensión es impar respecto de $0$ y también respecto de $L$:

$$
\widetilde f(-x)=-\widetilde f(x),
$$

$$
\widetilde f(L+x)=-\widetilde f(L-x).
$$


````python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,2,401)
desplegada = 0.25+t
fase = desplegada % 2
fisica = np.where(fase <= 1, fase, 2-fase)
fig, ax = plt.subplots(figsize=(7.4,3.8),constrained_layout=True)
ax.plot(t,desplegada,'--',color='0.45',label='Recorrido desplegado')
ax.plot(t,fisica,color='#00689d',lw=2.5,label='Posición en la cuerda')
ax.axhline(1, color='#aa452c',lw=1,label='Extremo derecho')
ax.scatter([0,0.75,1.5],[0.25,1,0.25],color='#00689d',zorder=4)
ax.set(xlabel='Tiempo',ylabel='Posición',title='Una reflexión seguida por una marca del pulso')
ax.legend()
plt.show()
````

**Lectura.** En la recta punteada la señal atraviesa la posición $1$ del problema desplegado. En la cuerda física (azul) regresa desde ese extremo. La ordenada mide posición, no amplitud; el cambio de signo se deduce de la extensión impar.

## **Proposición 2.6.4 (Regularidad de la extensión periódica).**

Si

$$
f\in C^2([0,L]),
$$

y

$$
f(0)=f(L)=f''(0)=f''(L)=0,
$$

entonces

$$
\widetilde f\in C^2(\mathbb R).
$$

Si

$$
g\in C^1([0,L]),
$$

y

$$
g(0)=g(L)=0,
$$

entonces

$$
\widetilde g\in C^1(\mathbb R).
$$

### Demostración

La regularidad se verifica en los puntos de reflexión $kL$. En cada extremo, la continuidad de la función exige la anulación del valor, mientras que la continuidad de la segunda derivada de una extensión impar exige la anulación de la segunda derivada.

$\square$

# 2.6.3. Solución por reflexiones múltiples

**Cómo leer el perfil extendido.** Cerca de $x=L$, los dos sumandos de d’Alembert que ven puntos simétricos tienen alturas opuestas en la pared y se cancelan. A un lado de ella dejan una señal reflejada. El valor que impones en la pared explica la elección de paridad de las copias.


## **Teorema 2.6.5 (Método de reflexión periódica para extremos fijos).**

Sean $\widetilde f$ y $\widetilde g$ las extensiones impares $2L$-periódicas de los datos.

Entonces

$$
\boxed{
u(x,t)
=
\frac12
\left[
\widetilde f(x-ct)+\widetilde f(x+ct)
\right]
+
\frac{1}{2c}
\int_{x-ct}^{x+ct}
\widetilde g(y)\,dy,
}
$$

restringida a

$$
0\leq x\leq L,
$$

satisface el problema de Dirichlet homogéneo.

### Demostración

La fórmula resuelve la ecuación de onda en toda la recta. La imparidad respecto de $0$ implica

$$
u(0,t)=0.
$$

La imparidad respecto de $L$ implica

$$
u(L,t)=0.
$$

En $t=0$ se recuperan $f$ y $g$ por restricción.

$\square$


## **Observación 2.6.6 (Interpretación geométrica).**

La extensión periódica reemplaza las reflexiones sucesivas por una propagación libre en una sucesión infinita de copias del intervalo.

Cada choque con un extremo fijo invierte el signo del pulso.

# 2.6.4. Escena Manim

La escena:

1. muestra primero el intervalo y el perfil inicial;
2. mantiene el perfil visible durante cuatro segundos;
3. introduce las condiciones de compatibilidad;
4. muestra la extensión periódica;
5. anima dos reflexiones sucesivas con inversión de signo.

El video no se compila en esta entrega.

Script:

`manim/reflexiones_en_intervalo.py`

Render directo:

```bash
manim -qh --fps 60 manim/reflexiones_en_intervalo.py ReflexionesEnIntervalo
```

````python
import numpy as np
from manim import *


class ReflexionesEnIntervalo(Scene):
    def construct(self):
        title = Text(
            "Onda en un intervalo con extremos fijos",
            font_size=44,
        ).to_edge(UP)

        problem = Text('uₜₜ-c²uₓₓ=0,     0<x<L,     u(0,t)=u(L,t)=0', font_size=30).scale(0.78).next_to(title, DOWN, buff=0.35)

        self.play(Write(title), Write(problem), run_time=2.4)
        self.wait(1.5)

        axes = Axes(
            x_range=[-0.3, 6.3, 1],
            y_range=[-1.2, 1.2, 0.5],
            x_length=10.8,
            y_length=4.4,
            axis_config={"include_tip": True},
        ).shift(DOWN * 0.9)

        labels = axes.get_axis_labels(Text('x', font_size=30), Text('u', font_size=30))

        L = 6.0
        left_wall = Line(axes.c2p(0, -1.15), axes.c2p(0, 1.15), stroke_width=6)
        right_wall = Line(axes.c2p(L, -1.15), axes.c2p(L, 1.15), stroke_width=6)

        self.play(
            Create(axes),
            Write(labels),
            Create(left_wall),
            Create(right_wall),
            run_time=2.5,
        )

        def bump(x, center=1.7, radius=0.72):
            q = (x-center)/radius
            if abs(q) >= 1:
                return 0.0
            return np.exp(-1/(1-q*q))/np.exp(-1)

        profile = axes.plot(
            lambda x: bump(x),
            x_range=[0, L],
        )
        profile_label = Text('f(x)', font_size=30).next_to(profile, UP, buff=0.2)

        self.play(Create(profile), Write(profile_label), run_time=2.2)

        # Perfil inicial visible antes de la evolución.
        self.wait(4.0)

        compatibility = Text('f(0)=f(L)=g(0)=g(L)=0', font_size=30).scale(0.78).to_corner(UL).shift(DOWN * 1.3)

        self.play(Write(compatibility), run_time=2.0)
        self.wait(2.0)

        extension = Text('f̃(x+2L)=f̃(x),     f̃(-x)=-f̃(x)', font_size=30).scale(0.72).to_edge(DOWN)

        self.play(Write(extension), run_time=2.4)
        self.wait(2.0)

        # Imágenes con signos opuestos: Dirichlet en ambos extremos.
        reloj = ValueTracker(0.)
        def valor(x):
            t = reloj.get_value()
            return sum(bump(x - t + 2*k*L) - bump(-x - t + 2*k*L)
                       for k in range(-2, 3))
        pulso = always_redraw(lambda: axes.plot(valor, x_range=[0,L,0.035], color=BLUE))
        self.remove(profile, profile_label)
        self.add(pulso)
        self.play(reloj.animate.set_value(12.), run_time=12., rate_func=linear)

        statement = Text('cada reflexión en un extremo fijo cambia el signo', font_size=30).scale(0.72).next_to(extension, UP, buff=0.35)

        self.play(Write(statement), run_time=2.0)
        self.wait(4.0)


# Render directo:
# manim -qh --fps 60 reflexiones_en_intervalo.py ReflexionesEnIntervalo
    
from IPython.display import Video, display
with tempconfig({"pixel_width": 2560, "pixel_height": 1440, "frame_rate": 120,
                 "write_to_movie": True, "format": "mp4", "preview": False}):
    escena = ReflexionesEnIntervalo()
    escena.render()
    video = escena.renderer.file_writer.movie_file_path
display(Video(str(video), embed=True))
````
